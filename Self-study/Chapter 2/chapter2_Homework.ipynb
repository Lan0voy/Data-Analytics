```python
import numpy as np
import pandas as pd

# Исторические данные
data = {
    "Year": [2004, 2005, 2006, 2007, 2008, 2009, 2010, 2011, 2012, 2013],
    "Num high school graduates": [25000, 25600, 24367, 25500, 28321, 28759, 25431, 24678, 24987, 25100],
    "USA Economics": [3, 3, 3, 3, 2, 1, 1, 1, 2, 2],
    "Comparable tuition fee": [2, 2, 3, 2, 3, 1, 1, 1, 2, 3],
    "Financial scholarship": [3, 1, 2, 2, 3, 3, 1, 1, 2, 2],
    "Financial aid available": [1, 2, 3, 3, 1, 3, 1, 2, 3, 3],
    "Number of open house visitors": [1100, 1005, 900, 980, 1212, 1220, 987, 899, 1488, 900],
    "Media advertisement investment": [13400, 12900, 8000, 8000, 7500, 10000, 8000, 7500, 11000, 11000],
    "Admission activities": [20, 15, 8, 11, 18, 19, 10, 8, 7, 10],
    "SAT cutoff": [1000, 980, 950, 1020, 1000, 1010, 950, 1000, 1000, 1000],
    "Enrollment": [200, 202, 188, 209, 232, 220, 188, 179, 190, 175]
}

df = pd.DataFrame(data)
df
```

```python
# Признаки X
X = df[
    [
        "Num high school graduates",
        "USA Economics",
        "Comparable tuition fee",
        "Financial scholarship",
        "Financial aid available",
        "Number of open house visitors",
        "Media advertisement investment",
        "Admission activities",
        "SAT cutoff"
    ]
].values

# Целевая переменная y
y = df["Enrollment"].values

# Добавляем столбец единиц для intercept b
X_with_intercept = np.column_stack([X, np.ones(len(X))])

# Находим коэффициенты методом наименьших квадратов
coefficients = np.linalg.lstsq(X_with_intercept, y, rcond=None)[0]

coefficients
```

```python
coef_names = [
    "Num high school graduates",
    "USA Economics",
    "Comparable tuition fee",
    "Financial scholarship",
    "Financial aid available",
    "Number of open house visitors",
    "Media advertisement investment",
    "Admission activities",
    "SAT cutoff",
    "Intercept"
]

coef_table = pd.DataFrame({
    "Coefficient": coef_names,
    "Value": coefficients
})

coef_table
```

```python
new_year = np.array([
    25500,  # Num high school graduates
    3,      # USA Economics
    3,      # Comparable tuition fee
    2,      # Financial scholarship
    2,      # Financial aid available
    1130,   # Number of open house visitors
    9000,   # Media advertisement investment
    20,     # Admission activities
    1000,   # SAT cutoff
    1       # Intercept
])

prediction = new_year @ coefficients

prediction
```

```python
print(f"Predicted number of first year enrollment: {prediction:.2f}")
```
